```python
from pathlib import Path
from IPython.display import display

from gmm_model_based_clustering import DEFAULT_DATASETS, run_full_experiment

base_dir = Path.cwd()
dataset_names = DEFAULT_DATASETS
output_dir = base_dir / "gmm_human_spine_outputs"

experiment = run_full_experiment(
    base_dir=base_dir,
    dataset_names=dataset_names,
    output_dir=output_dir,
    cluster_range=range(5, 6),
    covariance_types=("full", "tied", "diag", "spherical"),
    random_state=0,
)

print("Results saved to:", experiment["output_dir"])
```
