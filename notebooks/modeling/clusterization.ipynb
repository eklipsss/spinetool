# Clusterization

В ноутбуке есть 2 сценария:
1. `SphHarm + Volume`, `UMAP` до 3 размерностей.
2. `LF modulus`, без снижения размерности.

Основная оценка без разметки:
- количество кластеров;
- `silhouette score`;
- `Davies-Bouldin score`;
- `Calinski-Harabasz score`;
- размеры кластеров.


```python
from pathlib import Path

import plotly.io as pio
from IPython.display import display

from clusterization_utils import (
    choose_best_cluster_counts,
    load_lf_modulus_features,
    merge_sphharm_and_volume,
    plot_3d_embedding,
    plot_cluster_sweep,
    run_experiment,
    sweep_cluster_counts,
)

pio.renderers.default = "notebook_connected"
```

```python
archive_dir = Path(r"\Users\Student\Липс Екатерина\Spinetool - basic metric\Mouse_basal\z-corr")

sphharm_csv_path = archive_dir / "spherical_garmonics.csv"
metrics_csv_path = archive_dir / "metrics.csv"
# lf_modulus_csv_path = archive_dir / "light_field_zernike_moments.csv"

output_dir = archive_dir / "clusterization_outputs"
output_dir.mkdir(parents=True, exist_ok=True)

random_state = 0
significance_alpha = 0.05

cluster_range = range(2, 11)
n_clusters_sphharm_volume = 8
n_clusters_lf_modulus = 5

print("sphharm_csv_path:", sphharm_csv_path)
print("metrics_csv_path:", metrics_csv_path)
# print("lf_modulus_csv_path:", lf_modulus_csv_path)
print("output_dir:", output_dir)
```

## 1. SphHarm + Volume, UMAP (3D)

```python
sphharm_volume_features, sphharm_volume_metadata = merge_sphharm_and_volume(
    sphharm_csv_path=sphharm_csv_path,
    metrics_csv_path=metrics_csv_path,
)

print("SphHarm + Volume shape:", sphharm_volume_features.shape)
display(sphharm_volume_features.head())
```

```python
sphharm_volume_sweep = sweep_cluster_counts(
    features=sphharm_volume_features,
    cluster_counts=cluster_range,
    random_state=random_state,
    dim_reduction="umap",
    n_components=3,
)
sphharm_volume_best = choose_best_cluster_counts(sphharm_volume_sweep)

print(sphharm_volume_best)

display(sphharm_volume_sweep)
display(sphharm_volume_best)
plot_cluster_sweep(sphharm_volume_sweep, "SphHarm + Volume, UMAP(3D)").show()
```

```python
sphharm_volume_result = run_experiment(
    experiment_name="sphharm_volume_umap3",
    features=sphharm_volume_features,
    metadata=sphharm_volume_metadata,
    n_clusters=n_clusters_sphharm_volume,
    output_dir=output_dir,
    random_state=random_state,
    dim_reduction="umap",
    n_components=3,
    alpha=significance_alpha,
)
```

```python
display(sphharm_volume_result["unsupervised_summary"])
display(sphharm_volume_result["cluster_sizes"])

fig = plot_3d_embedding(sphharm_volume_result, "SphHarm + Volume, UMAP(3D)")
fig.show()
```

## 2. LF modulus, no dim. reduction

```python
if not lf_modulus_csv_path.exists():
    raise FileNotFoundError(
        "LF modulus file not found. Update lf_modulus_csv_path before running this cell."
    )

lf_modulus_features, lf_modulus_metadata = load_lf_modulus_features(lf_modulus_csv_path)

print("LF modulus shape:", lf_modulus_features.shape)
display(lf_modulus_features.head())
```

```python
lf_modulus_sweep = sweep_cluster_counts(
    features=lf_modulus_features,
    cluster_counts=cluster_range,
    random_state=random_state,
    dim_reduction=None,
    n_components=3,
)

display(lf_modulus_sweep)
plot_cluster_sweep(lf_modulus_sweep, "LF modulus, no reduction").show()
```

```python
lf_modulus_result = run_experiment(
    experiment_name="lf_modulus_no_reduction",
    features=lf_modulus_features,
    metadata=lf_modulus_metadata,
    n_clusters=n_clusters_lf_modulus,
    output_dir=output_dir,
    random_state=random_state,
    dim_reduction=None,
    n_components=3,
    alpha=significance_alpha,
)
```

```python
display(lf_modulus_result["unsupervised_summary"])
display(lf_modulus_result["cluster_sizes"])

print("Saved results to:", output_dir)
```
